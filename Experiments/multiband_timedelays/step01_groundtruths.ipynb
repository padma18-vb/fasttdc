```python
%load_ext autoreload
%autoreload 2
# imports
import pandas as pd
from astropy.cosmology import FlatLambdaCDM
from scipy.stats import norm
import sys
sys.path.insert(0, '/Users/smericks/Desktop/StrongLensing/padma_timedelays/fasttdc/')
import Utils.ground_truth_utils as gt_utils
```

Step #1: compute ground truth DDTs, time-delays, kinematics, etc.

```python
# assume a population over lambda_int, beta_ani and a ground truth cosmology

LAMBDA_INT_MU = 1.
LAMBDA_INT_SIGMA = 0.1
BETA_ANI_MU = 0.
BETA_ANI_SIGMA = 0.1

gt_cosmo = FlatLambdaCDM(H0=70.,Om0=0.3)
```

```python
# STEP 1: assign a lambda int and beta ani (ONLY ONE TIME)

# load in existing metadata and modify it
sample_df = pd.read_csv('DataVectors/truth_metadata.csv')

# only do this once (BE CAREFUL)
already_run = True
if not already_run:
    # add in lambda_int and beta_ani
    sample_df['lambda_int'] = norm.rvs(loc=LAMBDA_INT_MU,scale=LAMBDA_INT_SIGMA,size=len(sample_df))
    sample_df['beta_ani'] = norm.rvs(loc=BETA_ANI_MU,scale=BETA_ANI_SIGMA,size=len(sample_df))

    # re-save
    sample_df.to_csv('DataVectors/truth_metadata.csv',index=False)
    
```

```python
# STEP 2: pre-compute ground truth aperture kinematics (EXPENSIVE TIME-WISE)

# load in existing metadata and modify it
sample_df = pd.read_csv('DataVectors/truth_metadata.csv')

already_run = True
if not already_run: 
    # add a column with kinematics
    gt_utils.populate_truth_sigma_v_4MOST(sample_df,gt_cosmo)

    # re-save
    sample_df.to_csv('DataVectors/truth_metadata.csv',index=False)
```

```python
# STEP 3: pre-compute ground truth IFU kinematics (EXPENSIVE TIME-WISE)

# load in existing metadata and modify it
sample_df = pd.read_csv('DataVectors/truth_metadata.csv')

already_run = True
if not already_run: 
    # add a column with IFU kinematics
    gt_utils.populate_truth_sigma_v_IFU(sample_df,gt_cosmo)

    # re-save
    sample_df.to_csv('DataVectors/truth_metadata.csv',index=False)
```

```python
# STEP 4: compute ground truth Ddt, time-delays, fermat potentials

# load in existing metadata and modify it
sample_df = pd.read_csv('DataVectors/truth_metadata.csv')

already_run = True
if not already_run: 
    # add columns with fermat potentials, time-delays, DDTs
    gt_utils.populate_fermat_differences(sample_df)
    gt_utils.populate_truth_Ddt_timedelays(sample_df,gt_cosmo)

    # re-save
    sample_df.to_csv('DataVectors/truth_metadata.csv',index=False)
```
